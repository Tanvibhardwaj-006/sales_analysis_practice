```python
#1 Find total sales, total profit and total orders.
total_sales = df["Sales_INR"].sum()
total_profit = df["Profit_INR"].sum()
total_orders = df["Order_ID"].nunique()
print(f"Total Sales: {total_sales:,.2f}, Total Profit: {total_profit:,.2f}, Total Orders: {total_orders:,.2f}")
```

```python
#2 Find the top 5 products by total sales.
top_products = df.groupby("Product")["Sales_INR"].sum().sort_values(ascending=False).head(5)
print(top_products)
```

```python
#3 Calculate monthly sales and identify the best month.
monthly_sales = df.groupby(df['Order_Date'].dt.to_period('M'))['Sales_INR'].sum()
best_month = monthly_sales.sort_values(ascending=False).head(1)
print(monthly_sales)
print(best_month)
```

```python
#4 Compare sales across the five regions.
region_sales = df.groupby("Region")["Sales_INR"].sum()
print(region_sales)
```

```python
#5 Find the average order value.
average_order_value = df.groupby("Order_ID")["Sales_INR"].sum().mean()
print(f"Average Order Value: {average_order_value:,.2f}")
```

```python
#6 Calculate profit margin for each product category.
profit_margin = df.groupby("Category")["Profit_INR"].sum() / df.groupby("Category")["Sales_INR"].sum()
print(profit_margin)
```

```python
#7 Find the top 10 customers by total sales.
top_customers = df.groupby("Customer_Name")["Sales_INR"].sum().sort_values(ascending=False).head(10)
print(top_customers)
```

```python
#8 Compare payment modes by order count and sales.
payment_mode = df.groupby("Payment_Mode").agg(orders=("Order_ID", "count"), sales=("Sales_INR", "sum"))
print(payment_mode)
```

```python
#9 Compare Website, Mobile App and Store performance.
website_sales = df[df["Sales_Channel"] == "Website"]["Sales_INR"].sum()
mobile_app_sales = df[df["Sales_Channel"] == "Mobile App"]["Sales_INR"].sum()
store_sales = df[df["Sales_Channel"] == "Store"]["Sales_INR"].sum()
print(f"Website Sales: {website_sales:,.2f}, Mobile App Sales: {mobile_app_sales:,.2f}, Store Sales: {store_sales:,.2f}") 
```

```python
#10 Identify orders with discount greater than or equal to 15%.
high_discount_orders = df[df['Discount_Pct'] >= 0.15]
print(high_discount_orders)
```

```python
#12 Create a month-over-month sales growth calculation.
df.sort_values(by='Order_Date', inplace=True)
df['Month'] = df['Order_Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Sales_INR'].sum()
monthly_growth = monthly_sales.pct_change().fillna(0) * 100
print(monthly_growth)
```
