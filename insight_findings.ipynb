```python
import pandas as pd 
df = pd.read_excel('data_analyst_sales_practice_dataset.xlsx',sheet_name="Sales_Data")
df.head()
```

```python
#insight 2
top = df.groupby("Product")["Sales_INR"].sum().sort_values(ascending=False)
top.head(5).sum() / df["Sales_INR"].sum() * 100
```

```python
#insight 3
c = df.groupby("Category")[["Sales_INR", "Profit_INR"]].sum()
c["share_%"] = c["Sales_INR"] / c["Sales_INR"].sum() * 100
c["margin_%"] = c["Profit_INR"] / c["Sales_INR"] * 100
```

```python
#insight 4
hi = df[df["Discount_Pct"] >= 0.15]
lo = df[df["Discount_Pct"] < 0.15]
hi["Profit_INR"].sum() / hi["Sales_INR"].sum() * 100
```

```python
#insight 5
loss = df[df["Profit_INR"] < 0]
len(loss), loss["Profit_INR"].sum()
loss["Discount_Pct"].value_counts()
loss["Product"].value_counts()
```
